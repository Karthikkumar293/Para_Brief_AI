```python
# ============================================
# COMPLETE ENCODER-DECODER MODEL
# ============================================

import tensorflow as tf

# -------------------------------
# 1. INPUTS
# -------------------------------

encoder_inputs = tf.keras.Input(
    shape=(500,),
    dtype=tf.int32,
    name="encoder_inputs"
)

decoder_inputs = tf.keras.Input(
    shape=(66,),
    dtype=tf.int32,
    name="decoder_inputs"
)


# -------------------------------
# 2. SHARED EMBEDDING
# -------------------------------

embedding = tf.keras.layers.Embedding(
    input_dim=30000,
    output_dim=128,
    mask_zero=False,
    name="embedding"
)

encoder_embedded = embedding(encoder_inputs)
decoder_embedded = embedding(decoder_inputs)


# -------------------------------
# 3. ENCODER
# -------------------------------

encoder_lstm = tf.keras.layers.LSTM(
    256,
    return_sequences=True,
    return_state=True,
    name="encoder_lstm"
)

encoder_outputs, encoder_h, encoder_c = encoder_lstm(
    encoder_embedded
)


# -------------------------------
# 4. DECODER
# -------------------------------

decoder_lstm = tf.keras.layers.LSTM(
    256,
    return_sequences=True,
    return_state=True,
    name="decoder_lstm"
)

decoder_outputs, decoder_h, decoder_c = decoder_lstm(
    decoder_embedded,
    initial_state=[encoder_h, encoder_c]
)


# -------------------------------
# 5. ATTENTION
# -------------------------------

attention = tf.keras.layers.AdditiveAttention(
    name="attention"
)

context_vector = attention(
    [decoder_outputs, encoder_outputs]
)


# -------------------------------
# 6. COMBINE
# -------------------------------

decoder_combined = tf.keras.layers.Concatenate(
    axis=-1,
    name="decoder_combined"
)([
    decoder_outputs,
    context_vector
])


# -------------------------------
# 7. OUTPUT
# -------------------------------

output_layer = tf.keras.layers.Dense(
    30000,
    activation="softmax",
    name="output_layer"
)

final_outputs = output_layer(
    decoder_combined
)


# -------------------------------
# 8. COMPLETE MODEL
# -------------------------------

model = tf.keras.Model(
    inputs=[encoder_inputs, decoder_inputs],
    outputs=final_outputs
)


# -------------------------------
# 9. NEW OPTIMIZER
# -------------------------------

optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001
)


# -------------------------------
# 10. COMPILE
# -------------------------------

model.compile(
    optimizer=optimizer,
    loss=masked_loss
)


# -------------------------------
# 11. SUMMARY
# -------------------------------

model.summary()
```

```python
# ============================================================
# LOAD YOUR SAVED MODEL + TOKENIZER AND TEST SUMMARIZATION
# ============================================================

import tensorflow as tf
import pickle

# ------------------------------------------------------------
# 1. LOAD TOKENIZER
# ------------------------------------------------------------

with open("tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

# ------------------------------------------------------------
# 2. LOAD TRAINED MODEL
# ------------------------------------------------------------

model = tf.keras.models.load_model(
    "summarization_model.keras",
    compile=False
)

# ------------------------------------------------------------
# 3. TOKEN IDs
# ------------------------------------------------------------

START_TOKEN = tokenizer.word_index["<START>"]
END_TOKEN = tokenizer.word_index["<END>"]

print("Model loaded successfully ✅")
print("Tokenizer loaded successfully ✅")
print("START token:", START_TOKEN)
print("END token:", END_TOKEN)
print("Tokenizer vocabulary:", len(tokenizer.word_index))

# ------------------------------------------------------------
# 4. GENERATION FUNCTION
# ------------------------------------------------------------

def generate_summary(text):

    # Text → token IDs
    sequence = tokenizer.texts_to_sequences([text])

    # Pad document to 500 tokens
    document = tf.keras.preprocessing.sequence.pad_sequences(
        sequence,
        maxlen=500,
        padding="post",
        truncating="post"
    )

    # Start decoder with START
    generated = [START_TOKEN]

    for step in range(65):

        # Create decoder input
        decoder_input = tf.keras.preprocessing.sequence.pad_sequences(
            [generated],
            maxlen=66,
            padding="post",
            truncating="post"
        )

        # Predict
        predictions = model.predict(
            [document, decoder_input],
            verbose=0
        )

        # Probability for next token
        next_token = int(
            tf.argmax(
                predictions[0, len(generated) - 1]
            ).numpy()
        )

        # Stop if END token
        if next_token == END_TOKEN:
            break

        generated.append(next_token)

    # Token IDs → words
    words = []

    for token_id in generated:

        if token_id in [0, START_TOKEN, END_TOKEN]:
            continue

        word = tokenizer.index_word.get(
            token_id,
            "<UNK>"
        )

        # Don't show unknown tokens
        if word not in ["<OOV>", "<UNK>"]:
            words.append(word)

    return " ".join(words)


# ------------------------------------------------------------
# 5. PUT YOUR OWN PARAGRAPH HERE
# ------------------------------------------------------------

text = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""


# ------------------------------------------------------------
# 6. GENERATE SUMMARY
# ------------------------------------------------------------

summary = generate_summary(text)

print("\n========================================")
print("ORIGINAL DOCUMENT")
print("========================================")
print(text)

print("\n========================================")
print("GENERATED SUMMARY")
print("========================================")
print(summary)
```

```python
import tensorflow as tf
import pickle

# ============================================================
# 1. LOAD TRAINED MODEL
# ============================================================

model = tf.keras.models.load_model(
    "summarization_model_best.keras",
    compile=False
)

# ============================================================
# 2. LOAD TOKENIZER
# ============================================================

with open("tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

START_TOKEN = 29998
END_TOKEN = 29999

MAX_DOCUMENT_LENGTH = 500
MAX_SUMMARY_LENGTH = 66

print("Model loaded ✅")
print("Tokenizer loaded ✅")


# ============================================================
# 3. SUMMARY GENERATION
# ============================================================

def generate_summary(text):

    # Clean input
    text = text.lower()
    text = text.replace("\n", " ")

    # Convert text → token IDs
    sequence = tokenizer.texts_to_sequences([text])

    # Pad document to 500 tokens
    document = tf.keras.utils.pad_sequences(
        sequence,
        maxlen=MAX_DOCUMENT_LENGTH,
        padding="post",
        truncating="post"
    )

    # Start with START token
    generated = [START_TOKEN]

    # Generate one token at a time
    for _ in range(MAX_SUMMARY_LENGTH - 1):

        # Decoder input
        decoder_input = tf.keras.utils.pad_sequences(
            [generated],
            maxlen=MAX_SUMMARY_LENGTH,
            padding="post",
            truncating="post"
        )

        # Model prediction
        predictions = model.predict(
            [document, decoder_input],
            verbose=0
        )

        # Get prediction for current position
        position = len(generated) - 1

        probabilities = predictions[0, position]

        # Select highest-probability token
        next_token = int(tf.argmax(probabilities).numpy())

        # Stop if END token
        if next_token == END_TOKEN:
            break

        # Avoid padding token
        if next_token == 0:
            break

        generated.append(next_token)

    # Convert token IDs → words
    words = []

    for token_id in generated[1:]:
        word = tokenizer.index_word.get(token_id, "")

        if word and word != "<OOV>":
            words.append(word)

    return " ".join(words)


# ============================================================
# 4. TEST DOCUMENT
# ============================================================

text = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""


# ============================================================
# 5. GENERATE SUMMARY
# ============================================================

summary = generate_summary(text)

print("\n" + "=" * 60)
print("ORIGINAL DOCUMENT")
print("=" * 60)
print(text.strip())

print("\n" + "=" * 60)
print("GENERATED SUMMARY")
print("=" * 60)
print(summary)
```

```python
# ============================================================
# COMPILE MODEL
# ============================================================

loss_function = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=False,
    reduction="none"
)


def masked_loss(y_true, y_pred):

    loss = loss_function(
        y_true,
        y_pred
    )

    mask = tf.cast(
        tf.not_equal(y_true, 0),
        loss.dtype
    )

    loss = loss * mask

    return (
        tf.reduce_sum(loss)
        /
        tf.maximum(
            tf.reduce_sum(mask),
            1.0
        )
    )


model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss=masked_loss
)


print("Model compiled successfully.")
```

```python
# ============================================================
# FINAL SUMMARY GENERATION
# ============================================================

import tensorflow as tf
import numpy as np
import pickle

# ------------------------------------------------------------
# 1. LOAD TOKENIZER
# ------------------------------------------------------------

with open("tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

# ------------------------------------------------------------
# 2. LOAD BEST TRAINED MODEL
# ------------------------------------------------------------

model = tf.keras.models.load_model(
    "karthik_1model.best.keras",
    compile=False
)

# ------------------------------------------------------------
# 3. SPECIAL TOKENS
# ------------------------------------------------------------

START_TOKEN = tokenizer.word_index["<START>"]
END_TOKEN = tokenizer.word_index["<END>"]

print("Model loaded successfully ✅")
print("Parameters:", model.count_params())
print("START:", START_TOKEN)
print("END:", END_TOKEN)


# ============================================================
# 4. FUNCTION TO GENERATE SUMMARY
# ============================================================

def generate_summary(text, max_length=66):

    # --------------------------------------------------------
    # Tokenize document
    # --------------------------------------------------------

    tokens = tokenizer.texts_to_sequences([text])[0]

    # --------------------------------------------------------
    # Pad document
    # --------------------------------------------------------

    document = tf.keras.preprocessing.sequence.pad_sequences(
        [tokens],
        maxlen=500,
        padding="post",
        truncating="post",
        value=0
    )

    # --------------------------------------------------------
    # Start decoder
    # --------------------------------------------------------

    generated = [START_TOKEN]

    # --------------------------------------------------------
    # Generate one token at a time
    # --------------------------------------------------------

    for step in range(max_length - 1):

        decoder_input = np.zeros(
            (1, 66),
            dtype=np.int32
        )

        decoder_input[0, :len(generated)] = generated

        # ----------------------------------------------------
        # Predict
        # ----------------------------------------------------

        predictions = model.predict(
            {
                "encoder_inputs": document,
                "decoder_inputs": decoder_input
            },
            verbose=0
        )

        # ----------------------------------------------------
        # Current decoder position
        # ----------------------------------------------------

        position = len(generated) - 1

        probabilities = predictions[0, position].copy()

        # ----------------------------------------------------
        # Never generate START again
        # ----------------------------------------------------

        probabilities[START_TOKEN] = 0

        # ----------------------------------------------------
        # Never generate padding
        # ----------------------------------------------------

        probabilities[0] = 0

        # ----------------------------------------------------
        # Prevent immediate repetition
        # ----------------------------------------------------

        if len(generated) > 1:

            previous_token = generated[-1]

            probabilities[previous_token] = 0

        # ----------------------------------------------------
        # Select next token
        # ----------------------------------------------------

        next_token = int(
            np.argmax(probabilities)
        )

        # ----------------------------------------------------
        # Stop conditions
        # ----------------------------------------------------

        if next_token == END_TOKEN:
            break

        if next_token == 0:
            break

        generated.append(next_token)

    # ========================================================
    # 5. TOKEN IDs → WORDS
    # ========================================================

    words = []

    for token_id in generated[1:]:

        if token_id == END_TOKEN:
            break

        word = tokenizer.index_word.get(
            int(token_id),
            "<OOV>"
        )

        # Ignore unknown token
        if word == "<OOV>":
            continue

        words.append(word)

    return " ".join(words)


# ============================================================
# 6. PUT YOUR PARAGRAPH HERE
# ============================================================

paragraph = """
The city council has approved a major plan to improve local healthcare
services. The project will build two new clinics, upgrade existing
hospitals, and provide additional medical staff in areas with growing
populations. Officials said the improvements are expected to reduce
waiting times and make healthcare more accessible to residents.
"""


# ============================================================
# 7. GENERATE
# ============================================================

summary = generate_summary(paragraph)


# ============================================================
# 8. DISPLAY
# ============================================================

print("\n" + "=" * 60)
print("INPUT")
print("=" * 60)

print(paragraph.strip())

print("\n" + "=" * 60)
print("GENERATED SUMMARY")
print("=" * 60)

print(summary)
```

```python
import tensorflow as tf
import numpy as np
import pickle
import os

# ============================================================
# 1. MOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

MODEL_PATH = "/content/drive/MyDrive/summarization_project/summarization_model.keras"
TOKENIZER_PATH = "/content/drive/MyDrive/summarization_project/tokenizer.pkl"

# ============================================================
# 2. LOAD TOKENIZER
# ============================================================

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

# ============================================================
# 3. LOAD MODEL
# ============================================================

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

print("Model loaded")
print("Tokenizer loaded")
print("Model parameters:", model.count_params())

# ============================================================
# 4. SPECIAL TOKENS
# ============================================================

START_TOKEN = tokenizer.word_index.get("<START>")
END_TOKEN = tokenizer.word_index.get("<END>")
OOV_TOKEN = tokenizer.word_index.get("<OOV>")

print("START token:", START_TOKEN)
print("END token:", END_TOKEN)
print("OOV token:", OOV_TOKEN)
print("Vocabulary:", len(tokenizer.word_index))

# ============================================================
# 5. GET TOKENIZER INDEX
# ============================================================

index_word = tokenizer.index_word


# ============================================================
# 6. BEAM SEARCH SUMMARY GENERATION
# ============================================================

def generate_summary(
    text,
    max_input_length=500,
    max_summary_length=66,
    beam_width=3,
    length_penalty=0.7
):

    # --------------------------------------------------------
    # Convert article to token IDs
    # --------------------------------------------------------

    input_sequence = tokenizer.texts_to_sequences([text])[0]

    if len(input_sequence) == 0:
        return ""

    # --------------------------------------------------------
    # Pad article
    # --------------------------------------------------------

    encoder_input = tf.keras.preprocessing.sequence.pad_sequences(
        [input_sequence],
        maxlen=max_input_length,
        padding="post",
        truncating="post",
        value=0
    )

    # --------------------------------------------------------
    # Each beam:
    # (token_sequence, log_probability)
    # --------------------------------------------------------

    beams = [
        ([START_TOKEN], 0.0)
    ]

    completed = []

    # --------------------------------------------------------
    # Generate one token at a time
    # --------------------------------------------------------

    for step in range(max_summary_length - 1):

        candidates = []

        for sequence, score in beams:

            # If already ended, keep it
            if sequence[-1] == END_TOKEN:
                completed.append((sequence, score))
                continue

            # ------------------------------------------------
            # Decoder input
            # ------------------------------------------------

            decoder_input = np.zeros(
                (1, max_summary_length),
                dtype=np.int32
            )

            decoder_input[0, :len(sequence)] = sequence

            # ------------------------------------------------
            # Model prediction
            # ------------------------------------------------

            prediction = model.predict(
                {
                    "encoder_inputs": encoder_input,
                    "decoder_inputs": decoder_input
                },
                verbose=0
            )

            # ------------------------------------------------
            # Prediction for next position
            # ------------------------------------------------

            position = len(sequence) - 1

            probabilities = prediction[0, position].astype(
                np.float64
            )

            # ------------------------------------------------
            # Numerical safety
            # ------------------------------------------------

            probabilities = np.maximum(
                probabilities,
                1e-12
            )

            probabilities = probabilities / np.sum(probabilities)

            # ------------------------------------------------
            # Never generate padding
            # ------------------------------------------------

            probabilities[0] = 0.0

            # ------------------------------------------------
            # Never generate START again
            # ------------------------------------------------

            if START_TOKEN is not None:
                probabilities[START_TOKEN] = 0.0

            # ------------------------------------------------
            # Strongly discourage OOV
            # ------------------------------------------------

            if OOV_TOKEN is not None:
                probabilities[OOV_TOKEN] *= 0.01

            # ------------------------------------------------
            # Reduce repeated words
            # ------------------------------------------------

            if len(sequence) >= 2:

                last_token = sequence[-1]

                probabilities[last_token] *= 0.05

            # ------------------------------------------------
            # Prevent repeated bigrams
            # ------------------------------------------------

            if len(sequence) >= 2:

                previous_bigram = (
                    sequence[-2],
                    sequence[-1]
                )

                for token_id in range(len(probabilities)):

                    if (
                        len(sequence) >= 2
                        and (
                            sequence[-1],
                            token_id
                        ) == previous_bigram
                    ):
                        probabilities[token_id] *= 0.05

            # ------------------------------------------------
            # Normalize again
            # ------------------------------------------------

            total = np.sum(probabilities)

            if total <= 0:
                continue

            probabilities = probabilities / total

            # ------------------------------------------------
            # Select top candidate words
            # ------------------------------------------------

            top_indices = np.argsort(
                probabilities
            )[-beam_width:][::-1]

            for token_id in top_indices:

                probability = probabilities[token_id]

                if probability <= 0:
                    continue

                new_sequence = sequence + [
                    int(token_id)
                ]

                new_score = score + np.log(
                    probability
                )

                candidates.append(
                    (new_sequence, new_score)
                )

        # ----------------------------------------------------
        # Nothing left
        # ----------------------------------------------------

        if not candidates:
            break

        # ----------------------------------------------------
        # Keep best beams
        # ----------------------------------------------------

        def beam_score(item):

            sequence, score = item

            length = max(
                1,
                len(sequence) - 1
            )

            return score / (
                length ** length_penalty
            )

        candidates.sort(
            key=beam_score,
            reverse=True
        )

        beams = candidates[:beam_width]

        # ----------------------------------------------------
        # Stop if enough completed sequences
        # ----------------------------------------------------

        if len(completed) >= beam_width:
            break

    # ========================================================
    # SELECT FINAL SEQUENCE
    # ========================================================

    all_sequences = completed + beams

    if not all_sequences:
        return ""

    all_sequences.sort(
        key=beam_score,
        reverse=True
    )

    best_sequence = all_sequences[0][0]

    # ========================================================
    # CONVERT TOKEN IDs TO WORDS
    # ========================================================

    words = []

    for token_id in best_sequence:

        if token_id == START_TOKEN:
            continue

        if token_id == END_TOKEN:
            break

        if token_id == 0:
            continue

        word = index_word.get(
            int(token_id),
            None
        )

        if word is None:
            continue

        if word == "<OOV>":
            continue

        words.append(word)

    return " ".join(words)


# ============================================================
# 7. TEST PARAGRAPH
# ============================================================

text = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

# ============================================================
# 8. GENERATE SUMMARY
# ============================================================

summary = generate_summary(text)

print("=" * 60)
print("ORIGINAL")
print("=" * 60)
print(text.strip())

print("\n" + "=" * 60)
print("GENERATED SUMMARY")
print("=" * 60)
print(summary)
```

```python
# ============================================================
# COMPLETE REBUILD + TRAINING
# CORRECTED TOKENIZER / DATASET / MODEL
# ============================================================

import os
import re
import pickle
import numpy as np
import tensorflow as tf

from datasets import load_dataset
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import layers, Model


# ============================================================
# 1. CONFIGURATION
# ============================================================

VOCAB_SIZE = 30000
DOCUMENT_LENGTH = 500
SUMMARY_LENGTH = 66

EMBEDDING_DIM = 128
LSTM_UNITS = 256
BATCH_SIZE = 32

START_TOKEN = 2
END_TOKEN = 3
OOV_TOKEN = 1
PAD_TOKEN = 0


# ============================================================
# 2. LOAD XSUM DATASET
# ============================================================

print("Loading XSum dataset...")

xsum = load_dataset("EdinburghNLP/xsum")

train_data = xsum["train"].select(range(20000))
val_data = xsum["validation"].select(range(2000))

print("Train examples:", len(train_data))
print("Validation examples:", len(val_data))


# ============================================================
# 3. CLEAN TEXT
# ============================================================

def clean_text(text):
    text = text.lower()
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text)
    return text.strip()


train_documents = [
    clean_text(x["document"])
    for x in train_data
]

train_summaries = [
    clean_text(x["summary"])
    for x in train_data
]

val_documents = [
    clean_text(x["document"])
    for x in val_data
]

val_summaries = [
    clean_text(x["summary"])
    for x in val_data
]


# ============================================================
# 4. CREATE TOKENIZER
#
# IMPORTANT:
# 0 = PAD
# 1 = OOV
# 2 = START
# 3 = END
#
# We DO NOT shift OOV.
# ============================================================

tokenizer = Tokenizer(
    num_words=29998,
    oov_token="<OOV>",
    filters=""
)

tokenizer.fit_on_texts(
    train_documents
    + train_summaries
    + val_documents
    + val_summaries
)

print("Raw tokenizer vocabulary:", len(tokenizer.word_index))


# ============================================================
# 5. CORRECT TOKEN ID CONVERSION
# ============================================================

def correct_token_ids(sequence):

    corrected = []

    for token_id in sequence:

        token_id = int(token_id)

        # Keep OOV as 1
        if token_id == OOV_TOKEN:
            corrected.append(OOV_TOKEN)

        # Shift normal vocabulary IDs
        elif token_id >= 2:
            corrected.append(token_id + 2)

        # Safety
        else:
            corrected.append(token_id)

    return corrected


# ============================================================
# 6. DOCUMENT SEQUENCES
# ============================================================

train_document_sequences = [
    correct_token_ids(seq)
    for seq in tokenizer.texts_to_sequences(
        train_documents
    )
]

val_document_sequences = [
    correct_token_ids(seq)
    for seq in tokenizer.texts_to_sequences(
        val_documents
    )
]


# ============================================================
# 7. SUMMARY SEQUENCES
# ============================================================

train_summary_sequences = [
    correct_token_ids(seq)
    for seq in tokenizer.texts_to_sequences(
        train_summaries
    )
]

val_summary_sequences = [
    correct_token_ids(seq)
    for seq in tokenizer.texts_to_sequences(
        val_summaries
    )
]


# ============================================================
# 8. ADD START + END
# ============================================================

train_summary_sequences = [
    [START_TOKEN] + seq + [END_TOKEN]
    for seq in train_summary_sequences
]

val_summary_sequences = [
    [START_TOKEN] + seq + [END_TOKEN]
    for seq in val_summary_sequences
]


# ============================================================
# 9. PAD DOCUMENTS
# ============================================================

train_documents_padded = pad_sequences(
    train_document_sequences,
    maxlen=DOCUMENT_LENGTH,
    padding="post",
    truncating="post",
    value=PAD_TOKEN
)

val_documents_padded = pad_sequences(
    val_document_sequences,
    maxlen=DOCUMENT_LENGTH,
    padding="post",
    truncating="post",
    value=PAD_TOKEN
)


# ============================================================
# 10. CREATE DECODER INPUTS + TARGETS
# ============================================================

train_decoder_inputs = [
    seq[:-1]
    for seq in train_summary_sequences
]

train_decoder_targets = [
    seq[1:]
    for seq in train_summary_sequences
]

val_decoder_inputs = [
    seq[:-1]
    for seq in val_summary_sequences
]

val_decoder_targets = [
    seq[1:]
    for seq in val_summary_sequences
]


# ============================================================
# 11. PAD DECODER DATA
# ============================================================

train_decoder_inputs = pad_sequences(
    train_decoder_inputs,
    maxlen=SUMMARY_LENGTH,
    padding="post",
    truncating="post",
    value=PAD_TOKEN
)

train_decoder_targets = pad_sequences(
    train_decoder_targets,
    maxlen=SUMMARY_LENGTH,
    padding="post",
    truncating="post",
    value=PAD_TOKEN
)

val_decoder_inputs = pad_sequences(
    val_decoder_inputs,
    maxlen=SUMMARY_LENGTH,
    padding="post",
    truncating="post",
    value=PAD_TOKEN
)

val_decoder_targets = pad_sequences(
    val_decoder_targets,
    maxlen=SUMMARY_LENGTH,
    padding="post",
    truncating="post",
    value=PAD_TOKEN
)


# ============================================================
# 12. VERIFY DATA
# ============================================================

print()
print("=" * 60)
print("DATA SHAPES")
print("=" * 60)

print("Train documents:", train_documents_padded.shape)
print("Train decoder inputs:", train_decoder_inputs.shape)
print("Train decoder targets:", train_decoder_targets.shape)

print("Validation documents:", val_documents_padded.shape)
print("Validation decoder inputs:", val_decoder_inputs.shape)
print("Validation decoder targets:", val_decoder_targets.shape)

print()
print("SPECIAL TOKENS")
print("PAD:", PAD_TOKEN)
print("OOV:", OOV_TOKEN)
print("START:", START_TOKEN)
print("END:", END_TOKEN)


# ============================================================
# 13. CHECK TOKEN COLLISION
# ============================================================

print()
print("TOKEN COLLISION CHECK")

print("ID 0:", "PAD")
print("ID 1:", "OOV")
print("ID 2:", "START")
print("ID 3:", "END")

assert OOV_TOKEN != START_TOKEN
assert OOV_TOKEN != END_TOKEN
assert START_TOKEN != END_TOKEN

print("No special-token collision.")


# ============================================================
# 14. CREATE TF.DATA DATASETS
# ============================================================

train_dataset = tf.data.Dataset.from_tensor_slices(
    (
        {
            "encoder_inputs": train_documents_padded,
            "decoder_inputs": train_decoder_inputs
        },
        train_decoder_targets
    )
)

train_dataset = train_dataset.shuffle(
    len(train_documents_padded)
)

train_dataset = train_dataset.batch(
    BATCH_SIZE
)

train_dataset = train_dataset.prefetch(
    tf.data.AUTOTUNE
)


val_dataset = tf.data.Dataset.from_tensor_slices(
    (
        {
            "encoder_inputs": val_documents_padded,
            "decoder_inputs": val_decoder_inputs
        },
        val_decoder_targets
    )
)

val_dataset = val_dataset.batch(
    BATCH_SIZE
)

val_dataset = val_dataset.prefetch(
    tf.data.AUTOTUNE
)


# ============================================================
# 15. BUILD ENCODER
# ============================================================

encoder_inputs = tf.keras.Input(
    shape=(DOCUMENT_LENGTH,),
    dtype=tf.int32,
    name="encoder_inputs"
)

encoder_embedding = layers.Embedding(
    input_dim=VOCAB_SIZE,
    output_dim=EMBEDDING_DIM,
    mask_zero=True,
    name="encoder_embedding"
)

encoder_embedded = encoder_embedding(
    encoder_inputs
)

encoder_lstm = layers.LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    use_cudnn=False,
    name="encoder_lstm"
)

encoder_outputs, encoder_h, encoder_c = encoder_lstm(
    encoder_embedded
)


# ============================================================
# 16. BUILD DECODER
# ============================================================

decoder_inputs = tf.keras.Input(
    shape=(SUMMARY_LENGTH,),
    dtype=tf.int32,
    name="decoder_inputs"
)

decoder_embedding = layers.Embedding(
    input_dim=VOCAB_SIZE,
    output_dim=EMBEDDING_DIM,
    mask_zero=True,
    name="decoder_embedding"
)

decoder_embedded = decoder_embedding(
    decoder_inputs
)

decoder_lstm = layers.LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    use_cudnn=False,
    name="decoder_lstm"
)

decoder_outputs, _, _ = decoder_lstm(
    decoder_embedded,
    initial_state=[
        encoder_h,
        encoder_c
    ]
)


# ============================================================
# 17. ADDITIVE ATTENTION
# ============================================================

attention = layers.AdditiveAttention(
    name="attention"
)

attention_output = attention(
    [
        decoder_outputs,
        encoder_outputs
    ]
)


# ============================================================
# 18. CONCATENATE
# ============================================================

combined = layers.Concatenate(
    axis=-1,
    name="decoder_attention_concat"
)(
    [
        decoder_outputs,
        attention_output
    ]
)


# ============================================================
# 19. OUTPUT
# ============================================================

outputs = layers.Dense(
    VOCAB_SIZE,
    activation="softmax",
    name="output"
)(
    combined
)


# ============================================================
# 20. CREATE MODEL
# ============================================================

model = Model(
    inputs=[
        encoder_inputs,
        decoder_inputs
    ],
    outputs=outputs
)


print()
print("=" * 60)
print("MODEL")
print("=" * 60)

print("Input shapes:", model.input_shape)
print("Output shape:", model.output_shape)
print("Parameters:", model.count_params())


# ============================================================
# 21. MASKED LOSS
# ============================================================

loss_function = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=False,
    reduction="none"
)


def masked_loss(y_true, y_pred):

    loss = loss_function(
        y_true,
        y_pred
    )

    mask = tf.cast(
        tf.not_equal(y_true, PAD_TOKEN),
        loss.dtype
    )

    loss = loss * mask

    return (
        tf.reduce_sum(loss)
        /
        tf.maximum(
            tf.reduce_sum(mask),
            1.0
        )
    )


# ============================================================
# 22. COMPILE
# ============================================================

optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001
)

model.compile(
    optimizer=optimizer,
    loss=masked_loss
)

print()
print("Model compiled successfully.")


# ============================================================
# 23. TEST ONE BATCH BEFORE TRAINING
# ============================================================

batch_inputs, batch_targets = next(
    iter(train_dataset)
)

predictions = model(
    batch_inputs,
    training=False
)

test_loss = masked_loss(
    batch_targets,
    predictions
)

print()
print("=" * 60)
print("ONE BATCH TEST")
print("=" * 60)

print("Documents:", batch_inputs["encoder_inputs"].shape)
print("Decoder inputs:", batch_inputs["decoder_inputs"].shape)
print("Targets:", batch_targets.shape)
print("Predictions:", predictions.shape)
print("Initial loss:", float(test_loss))


# ============================================================
# 24. TRAIN
# ============================================================

checkpoint_path = "/content/karthik_1model.best.keras"

print()
print("=" * 60)
print("STARTING TRAINING")
print("=" * 60)

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=8,
    callbacks=[
        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=2,
            restore_best_weights=True,
            verbose=1
        ),
        tf.keras.callbacks.ModelCheckpoint(
            checkpoint_path,
            monitor="val_loss",
            save_best_only=True,
            verbose=1
        )
    ]
)


# ============================================================
# 25. SAVE BEST MODEL IN COLAB
# ============================================================

model.save(
    checkpoint_path
)

print()
print("=" * 60)
print("COLAB MODEL SAVED")
print("=" * 60)

print(checkpoint_path)
print(
    "Size:",
    round(
        os.path.getsize(checkpoint_path)
        / (1024 * 1024),
        2
    ),
    "MB"
)


# ============================================================
# 26. SAVE TOKENIZER IN COLAB
# ============================================================

tokenizer_path = "/content/tokenizer_corrected.pkl"

with open(tokenizer_path, "wb") as f:
    pickle.dump(tokenizer, f)

print()
print("Tokenizer saved:")
print(tokenizer_path)


# ============================================================
# 27. COPY BOTH TO GOOGLE DRIVE
# ============================================================

from google.colab import drive
import shutil

drive.mount("/content/drive")

drive_folder = (
    "/content/drive/MyDrive/"
    "summarization_project"
)

os.makedirs(
    drive_folder,
    exist_ok=True
)

drive_model = os.path.join(
    drive_folder,
    "karthik_1model.best.keras"
)

drive_tokenizer = os.path.join(
    drive_folder,
    "tokenizer_corrected.pkl"
)

shutil.copy2(
    checkpoint_path,
    drive_model
)

shutil.copy2(
    tokenizer_path,
    drive_tokenizer
)


# ============================================================
# 28. FINAL CHECK
# ============================================================

print()
print("=" * 60)
print("BACKUP COMPLETE")
print("=" * 60)

print("Model in Colab:", os.path.exists(checkpoint_path))
print("Tokenizer in Colab:", os.path.exists(tokenizer_path))

print("Model in Drive:", os.path.exists(drive_model))
print("Tokenizer in Drive:", os.path.exists(drive_tokenizer))

print()
print("Drive model:")
print(drive_model)

print()
print("Drive tokenizer:")
print(drive_tokenizer)
```

```python
# ============================================================
# GENERATE SUMMARY USING THE NEW MODEL
# ============================================================

import tensorflow as tf
import numpy as np
import pickle
import re

MODEL_PATH = "/content/drive/MyDrive/summarization_project/karthik_1model.best.keras"
TOKENIZER_PATH = "/content/drive/MyDrive/summarization_project/tokenizer_corrected.pkl"

DOCUMENT_LENGTH = 500
SUMMARY_LENGTH = 66

START_TOKEN = 2
END_TOKEN = 3
OOV_TOKEN = 1
PAD_TOKEN = 0

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

# ------------------------------------------------------------
# LOAD MATCHING TOKENIZER
# ------------------------------------------------------------

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

index_word = tokenizer.index_word

print("Model loaded")
print("Tokenizer loaded")
print("Model output:", model.output_shape)
print("Tokenizer vocabulary:", len(tokenizer.word_index))


# ============================================================
# SUMMARY GENERATION
# ============================================================

def generate_summary(text):

    text = text.lower()
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text).strip()

    # --------------------------------------------------------
    # Convert document to tokens
    # --------------------------------------------------------

    sequence = tokenizer.texts_to_sequences([text])[0]

    # --------------------------------------------------------
    # Correct token IDs
    # --------------------------------------------------------

    corrected_sequence = []

    for token_id in sequence:

        token_id = int(token_id)

        if token_id == OOV_TOKEN:
            corrected_sequence.append(OOV_TOKEN)

        elif token_id >= 2:
            corrected_sequence.append(token_id + 2)

        else:
            corrected_sequence.append(token_id)

    # --------------------------------------------------------
    # Pad document
    # --------------------------------------------------------

    encoder_input = tf.keras.preprocessing.sequence.pad_sequences(
        [corrected_sequence],
        maxlen=DOCUMENT_LENGTH,
        padding="post",
        truncating="post",
        value=PAD_TOKEN
    )

    # --------------------------------------------------------
    # Start decoder
    # --------------------------------------------------------

    generated = [START_TOKEN]

    # --------------------------------------------------------
    # Generate one word at a time
    # --------------------------------------------------------

    for step in range(SUMMARY_LENGTH - 1):

        decoder_input = np.zeros(
            (1, SUMMARY_LENGTH),
            dtype=np.int32
        )

        decoder_input[0, :len(generated)] = generated

        prediction = model.predict(
            {
                "encoder_inputs": encoder_input,
                "decoder_inputs": decoder_input
            },
            verbose=0
        )

        position = len(generated) - 1

        probabilities = prediction[
            0,
            position,
            :
        ]

        # Never select padding
        probabilities[PAD_TOKEN] = 0

        # Never select START again
        probabilities[START_TOKEN] = 0

        # Avoid OOV
        probabilities[OOV_TOKEN] = 0

        # Avoid immediate repetition
        if len(generated) > 1:
            probabilities[generated[-1]] *= 0.05

        next_token = int(
            np.argmax(probabilities)
        )

        # END token
        if next_token == END_TOKEN:
            break

        generated.append(next_token)

    # --------------------------------------------------------
    # Convert tokens to words
    # --------------------------------------------------------

    words = []

    for token_id in generated:

        if token_id in [
            PAD_TOKEN,
            START_TOKEN,
            END_TOKEN,
            OOV_TOKEN
        ]:
            continue

        word = index_word.get(
            int(token_id)
        )

        if word is not None:
            words.append(word)

    return " ".join(words)


# ============================================================
# TEST
# ============================================================

text = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

summary = generate_summary(text)

print()
print("=" * 60)
print("ORIGINAL")
print("=" * 60)
print(text.strip())

print()
print("=" * 60)
print("GENERATED SUMMARY")
print("=" * 60)
print(summary)
```

```python
import tensorflow as tf
import numpy as np
import pickle
import re

# ============================================================
# LOAD MODEL + TOKENIZER
# ============================================================

MODEL_PATH = "/content/drive/MyDrive/summarization_project/karthik_1model.best.keras"
TOKENIZER_PATH = "/content/drive/MyDrive/summarization_project/tokenizer_corrected.pkl"

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

print("Model loaded")
print("Tokenizer loaded")
print("Model output:", model.output_shape)
print("Tokenizer vocabulary:", len(tokenizer.word_index))


# ============================================================
# IMPORTANT TOKEN IDs
# ============================================================

PAD_ID = 0
OOV_ID = 1

# These are the special IDs used by the notebook/model
START_ID = 29998
END_ID = 29999

print()
print("PAD:", PAD_ID)
print("OOV:", OOV_ID)
print("START:", START_ID)
print("END:", END_ID)


# ============================================================
# TOKENIZE DOCUMENT
# ============================================================

def prepare_document(text):

    text = text.lower()
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text).strip()

    sequence = tokenizer.texts_to_sequences([text])[0]

    # The trained model used the shifted vocabulary.
    # Keep OOV as 1 and shift normal tokenizer IDs.
    corrected = []

    for token_id in sequence:

        token_id = int(token_id)

        if token_id == OOV_ID:
            corrected.append(OOV_ID)
        else:
            corrected.append(token_id + 2)

    corrected = corrected[:500]

    if len(corrected) < 500:
        corrected += [PAD_ID] * (500 - len(corrected))

    return np.array(
        [corrected],
        dtype=np.int32
    )


# ============================================================
# GENERATE SUMMARY
# ============================================================

def generate_summary(text):

    encoder_input = prepare_document(text)

    generated = [START_ID]

    for step in range(65):

        decoder_input = np.zeros(
            (1, 66),
            dtype=np.int32
        )

        current_length = min(
            len(generated),
            66
        )

        decoder_input[0, :current_length] = \
            generated[:current_length]

        predictions = model.predict(
            {
                "encoder_inputs": encoder_input,
                "decoder_inputs": decoder_input
            },
            verbose=0
        )

        position = current_length - 1

        probabilities = predictions[
            0,
            position,
            :
        ].copy()

        # Do not generate padding
        probabilities[PAD_ID] = 0

        # Do not generate START repeatedly
        probabilities[START_ID] = 0

        next_id = int(
            np.argmax(probabilities)
        )

        if next_id == END_ID:
            break

        generated.append(next_id)

    # ========================================================
    # CONVERT MODEL IDS BACK TO TOKENIZER IDS
    # ========================================================

    words = []

    for model_id in generated[1:]:

        model_id = int(model_id)

        if model_id == END_ID:
            break

        if model_id == PAD_ID:
            continue

        if model_id == OOV_ID:
            words.append("<OOV>")
            continue

        # Reverse the +2 shift used during training
        tokenizer_id = model_id - 2

        if tokenizer_id <= 0:
            continue

        word = tokenizer.index_word.get(
            tokenizer_id
        )

        if word is not None:
            words.append(word)

    return " ".join(words)


# ============================================================
# TEST PARAGRAPH
# ============================================================

text = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

summary = generate_summary(text)

print()
print("=" * 60)
print("ORIGINAL")
print("=" * 60)
print(text.strip())

print()
print("=" * 60)
print("GENERATED SUMMARY")
print("=" * 60)
print(summary)
```

```python
# ============================================================
# CELL 2: CLEAN 50K MODEL + TRAINING
# NO CUdNN MASKING ERROR
# ============================================================

import tensorflow as tf
import numpy as np

VOCAB_SIZE = 50000
EMBED_DIM = 128
LATENT_DIM = 256

MAX_DOCUMENT_LENGTH = 500
MAX_SUMMARY_LENGTH = 66

PAD_ID = 0

# ============================================================
# MASKED LOSS
# ============================================================

def masked_loss(y_true, y_pred):

    loss = tf.keras.losses.sparse_categorical_crossentropy(
        y_true,
        y_pred
    )

    mask = tf.cast(
        tf.not_equal(y_true, PAD_ID),
        loss.dtype
    )

    loss = loss * mask

    return tf.reduce_sum(loss) / (
        tf.reduce_sum(mask) + 1e-7
    )


# ============================================================
# INPUTS
# ============================================================

encoder_inputs = tf.keras.Input(
    shape=(MAX_DOCUMENT_LENGTH,),
    dtype=tf.int32,
    name="encoder_inputs"
)

decoder_inputs = tf.keras.Input(
    shape=(MAX_SUMMARY_LENGTH,),
    dtype=tf.int32,
    name="decoder_inputs"
)


# ============================================================
# EMBEDDING
# IMPORTANT: mask_zero=False
# This avoids the cuDNN mask error.
# ============================================================

embedding = tf.keras.layers.Embedding(
    input_dim=VOCAB_SIZE,
    output_dim=EMBED_DIM,
    mask_zero=False,
    name="embedding"
)

encoder_embedded = embedding(encoder_inputs)
decoder_embedded = embedding(decoder_inputs)


# ============================================================
# ENCODER
# ============================================================

encoder_lstm = tf.keras.layers.LSTM(
    LATENT_DIM,
    return_sequences=True,
    return_state=True,
    name="encoder_lstm"
)

encoder_outputs, state_h, state_c = encoder_lstm(
    encoder_embedded
)


# ============================================================
# DECODER
# ============================================================

decoder_lstm = tf.keras.layers.LSTM(
    LATENT_DIM,
    return_sequences=True,
    return_state=True,
    name="decoder_lstm"
)

decoder_outputs, _, _ = decoder_lstm(
    decoder_embedded,
    initial_state=[state_h, state_c]
)


# ============================================================
# ATTENTION
# ============================================================

attention = tf.keras.layers.AdditiveAttention(
    name="attention"
)

context = attention(
    [
        decoder_outputs,
        encoder_outputs
    ]
)


# ============================================================
# COMBINE DECODER + ATTENTION
# ============================================================

decoder_combined = tf.keras.layers.Concatenate(
    axis=-1,
    name="decoder_combined"
)(
    [
        decoder_outputs,
        context
    ]
)


# ============================================================
# OUTPUT
# ============================================================

outputs = tf.keras.layers.Dense(
    VOCAB_SIZE,
    activation="softmax",
    name="output_layer"
)(
    decoder_combined
)


# ============================================================
# CREATE MODEL
# ============================================================

model = tf.keras.Model(
    inputs=[
        encoder_inputs,
        decoder_inputs
    ],
    outputs=outputs
)


# ============================================================
# COMPILE
# ============================================================

optimizer = tf.keras.optimizers.Adam(
    learning_rate=1e-3
)

model.compile(
    optimizer=optimizer,
    loss=masked_loss
)

print("Model compiled successfully.")
print("Output shape:", model.output_shape)
print("Parameters:", model.count_params())


# ============================================================
# CHECK DATA BEFORE TRAINING
# ============================================================

print()
print("Checking training dataset...")

sample = next(iter(train_dataset))

if isinstance(sample, tuple):
    x_sample, y_sample = sample

    print("Input type: tuple")
    print("Number of inputs:", len(x_sample) if isinstance(x_sample, (tuple, list)) else 1)

    if isinstance(x_sample, (tuple, list)):
        for i, x in enumerate(x_sample):
            print("Input", i, "shape:", x.shape)
    else:
        print("Input shape:", x_sample.shape)

    print("Target shape:", y_sample.shape)

else:
    print("Dataset sample type:", type(sample))


# ============================================================
# TRAIN
# ============================================================

print()
print("Starting training...")
print("Vocabulary:", VOCAB_SIZE)
print("Document length:", MAX_DOCUMENT_LENGTH)
print("Summary length:", MAX_SUMMARY_LENGTH)

checkpoint_path = "/content/karthik_50k_best.keras"

callbacks = [

    tf.keras.callbacks.ModelCheckpoint(
        checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True,
        verbose=1
    )
]

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=8,
    callbacks=callbacks
)

print()
print("============================================================")
print("TRAINING COMPLETED")
print("============================================================")
print("Best model:", checkpoint_path)
```

```python
import tensorflow as tf
import numpy as np
import pickle
import re
import os
from google.colab import drive

drive.mount("/content/drive")

PROJECT = "/content/drive/MyDrive/summarization_project"

MODEL_PATH = os.path.join(
    PROJECT,
    "karthik_50k_best.keras"
)

TOKENIZER_PATH = os.path.join(
    PROJECT,
    "tokenizer_corrected.pkl"
)

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

print("Model loaded:", os.path.exists(MODEL_PATH))
print("Tokenizer loaded:", os.path.exists(TOKENIZER_PATH))
print("Model output:", model.output_shape)
print("Tokenizer num_words:", tokenizer.num_words)
print("Tokenizer vocabulary:", len(tokenizer.word_index))

PAD_ID = 0
OOV_ID = 1
START_ID = 2
END_ID = 3

MAX_DOCUMENT_LENGTH = 500
MAX_SUMMARY_LENGTH = 66

def prepare_document(text):

    text = text.lower()
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text).strip()

    raw_ids = tokenizer.texts_to_sequences([text])[0]

    model_ids = []

    for token_id in raw_ids:

        token_id = int(token_id)

        if token_id == OOV_ID:
            model_ids.append(OOV_ID)
        else:
            model_ids.append(token_id + 2)

    model_ids = model_ids[:MAX_DOCUMENT_LENGTH]

    if len(model_ids) < MAX_DOCUMENT_LENGTH:
        model_ids += [PAD_ID] * (
            MAX_DOCUMENT_LENGTH - len(model_ids)
        )

    return np.array(
        [model_ids],
        dtype=np.int32
    )

def model_id_to_word(model_id):

    model_id = int(model_id)

    if model_id in [PAD_ID, START_ID, END_ID]:
        return None

    if model_id == OOV_ID:
        return "<OOV>"

    tokenizer_id = model_id - 2

    if tokenizer_id <= 0:
        return "<OOV>"

    return tokenizer.index_word.get(
        tokenizer_id,
        "<OOV>"
    )

def generate_summary(text):

    encoder_input = prepare_document(text)

    generated = [START_ID]

    for step in range(MAX_SUMMARY_LENGTH - 1):

        decoder_input = np.zeros(
            (1, MAX_SUMMARY_LENGTH),
            dtype=np.int32
        )

        decoder_input[0, :len(generated)] = generated

        predictions = model.predict(
            {
                "encoder_inputs": encoder_input,
                "decoder_inputs": decoder_input
            },
            verbose=0
        )

        position = len(generated) - 1

        probabilities = predictions[
            0,
            position,
            :
        ].copy()

        probabilities[PAD_ID] = -np.inf
        probabilities[START_ID] = -np.inf

        next_id = int(
            np.argmax(probabilities)
        )

        if next_id == END_ID:
            break

        generated.append(next_id)

    words = []

    for model_id in generated[1:]:

        if model_id == END_ID:
            break

        word = model_id_to_word(model_id)

        if word is not None:
            words.append(word)

    return " ".join(words), generated


text = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

summary, generated_ids = generate_summary(text)

print()
print("ORIGINAL:")
print(text.strip())

print()
print("GENERATED SUMMARY:")
print(summary)

print()
print("GENERATED MODEL IDS:")
print(generated_ids)
```

```python
# ============================================================
# CLEAN 50K SUMMARIZATION REBUILD
# XSum + LSTM Encoder/Decoder + Attention
# ============================================================

import os
import re
import pickle
import numpy as np
import tensorflow as tf
from datasets import load_dataset

# ------------------------------------------------------------
# 1. SETTINGS
# ------------------------------------------------------------

VOCAB_SIZE = 50000

PAD_ID   = 0
OOV_ID   = 1
START_ID = 2
END_ID   = 3

ENCODER_LEN = 500
DECODER_LEN = 66

EMBED_DIM = 128
LSTM_UNITS = 256
BATCH_SIZE = 32
EPOCHS = 3

DRIVE_DIR = "/content/drive/MyDrive/summarization_project"
os.makedirs(DRIVE_DIR, exist_ok=True)

MODEL_PATH = os.path.join(
    DRIVE_DIR,
    "summarization_50k_fixed.keras"
)

TOKENIZER_PATH = os.path.join(
    DRIVE_DIR,
    "tokenizer_50k_fixed.pkl"
)

# ------------------------------------------------------------
# 2. LOAD XSUM
# ------------------------------------------------------------

print("Loading XSum...")

dataset = load_dataset("EdinburghNLP/xsum")

train_data = dataset["train"].select(range(20000))
val_data   = dataset["validation"].select(range(2000))

train_docs = [
    str(x["document"])
    for x in train_data
]

train_summaries = [
    str(x["summary"])
    for x in train_data
]

val_docs = [
    str(x["document"])
    for x in val_data
]

val_summaries = [
    str(x["summary"])
    for x in val_data
]

print("Train:", len(train_docs))
print("Validation:", len(val_docs))

# ------------------------------------------------------------
# 3. TEXT CLEANING
# ------------------------------------------------------------

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()

train_docs_clean = [
    clean_text(x)
    for x in train_docs
]

train_summaries_clean = [
    clean_text(x)
    for x in train_summaries
]

val_docs_clean = [
    clean_text(x)
    for x in val_docs
]

val_summaries_clean = [
    clean_text(x)
    for x in val_summaries
]

# ------------------------------------------------------------
# 4. BUILD FRESH TOKENIZER
#
# Keras raw IDs:
# 0 = padding
# 1 = OOV
# 2... = normal words
#
# We convert them to model IDs:
# 0 = PAD
# 1 = OOV
# 2 = START
# 3 = END
# 4... = normal words
#
# Therefore raw tokenizer IDs 2...49997
# become model IDs 4...49999.
# ------------------------------------------------------------

tokenizer = tf.keras.preprocessing.text.Tokenizer(
    num_words=49998,
    oov_token="<OOV>",
    filters='!"#$%&()*+,-./:;=?@[\\]^_`{|}~\t\n'
)

print("Fitting new tokenizer...")

tokenizer.fit_on_texts(
    train_docs_clean + train_summaries_clean
)

print("Raw tokenizer vocabulary:",
      len(tokenizer.word_index))

print("Tokenizer num_words:",
      tokenizer.num_words)

# ------------------------------------------------------------
# 5. CONVERT RAW TOKENIZER IDS TO MODEL IDS
# ------------------------------------------------------------

def convert_ids(raw_ids):
    result = []

    for x in raw_ids:
        x = int(x)

        if x == 0:
            result.append(PAD_ID)

        elif x == 1:
            result.append(OOV_ID)

        else:
            new_id = x + 2

            if new_id < VOCAB_SIZE:
                result.append(new_id)
            else:
                result.append(OOV_ID)

    return result

# ------------------------------------------------------------
# 6. DOCUMENT ENCODING
# ------------------------------------------------------------

def encode_documents(texts):
    raw = tokenizer.texts_to_sequences(texts)

    converted = [
        convert_ids(x[:ENCODER_LEN])
        for x in raw
    ]

    return tf.keras.utils.pad_sequences(
        converted,
        maxlen=ENCODER_LEN,
        padding="post",
        truncating="post",
        value=PAD_ID
    ).astype("int32")

# ------------------------------------------------------------
# 7. SUMMARY ENCODING
# ------------------------------------------------------------

def encode_summaries(texts):
    raw = tokenizer.texts_to_sequences(texts)

    converted = []

    for ids in raw:

        ids = convert_ids(ids)

        # Leave room for START and END
        ids = ids[:DECODER_LEN - 2]

        full = (
            [START_ID]
            + ids
            + [END_ID]
        )

        converted.append(full)

    return tf.keras.utils.pad_sequences(
        converted,
        maxlen=DECODER_LEN,
        padding="post",
        truncating="post",
        value=PAD_ID
    ).astype("int32")

# ------------------------------------------------------------
# 8. CREATE TRAINING ARRAYS
# ------------------------------------------------------------

print("Encoding training data...")

X_train = encode_documents(train_docs_clean)

Y_full_train = encode_summaries(
    train_summaries_clean
)

decoder_inputs_train = Y_full_train[:, :-1]

decoder_targets_train = Y_full_train[:, 1:]

# Make decoder length exactly 66
decoder_inputs_train = np.pad(
    decoder_inputs_train,
    ((0, 0), (0, 1)),
    constant_values=PAD_ID
)

decoder_targets_train = np.pad(
    decoder_targets_train,
    ((0, 0), (0, 1)),
    constant_values=PAD_ID
)

decoder_inputs_train = decoder_inputs_train[:, :DECODER_LEN]
decoder_targets_train = decoder_targets_train[:, :DECODER_LEN]

print("Encoder:", X_train.shape)
print("Decoder input:", decoder_inputs_train.shape)
print("Decoder target:", decoder_targets_train.shape)

# ------------------------------------------------------------
# 9. VALIDATION DATA
# ------------------------------------------------------------

print("Encoding validation data...")

X_val = encode_documents(val_docs_clean)

Y_full_val = encode_summaries(
    val_summaries_clean
)

decoder_inputs_val = Y_full_val[:, :-1]

decoder_targets_val = Y_full_val[:, 1:]

decoder_inputs_val = np.pad(
    decoder_inputs_val,
    ((0, 0), (0, 1)),
    constant_values=PAD_ID
)

decoder_targets_val = np.pad(
    decoder_targets_val,
    ((0, 0), (0, 1)),
    constant_values=PAD_ID
)

decoder_inputs_val = decoder_inputs_val[:, :DECODER_LEN]
decoder_targets_val = decoder_targets_val[:, :DECODER_LEN]

print("Validation encoder:", X_val.shape)
print("Validation decoder:", decoder_inputs_val.shape)
print("Validation targets:", decoder_targets_val.shape)

# ------------------------------------------------------------
# 10. TF DATASETS
# ------------------------------------------------------------

train_dataset = tf.data.Dataset.from_tensor_slices(
    (
        {
            "encoder_inputs": X_train,
            "decoder_inputs": decoder_inputs_train
        },
        decoder_targets_train
    )
)

train_dataset = (
    train_dataset
    .shuffle(20000)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

val_dataset = tf.data.Dataset.from_tensor_slices(
    (
        {
            "encoder_inputs": X_val,
            "decoder_inputs": decoder_inputs_val
        },
        decoder_targets_val
    )
)

val_dataset = (
    val_dataset
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

# ------------------------------------------------------------
# 11. BUILD MODEL
# ------------------------------------------------------------

encoder_inputs = tf.keras.Input(
    shape=(ENCODER_LEN,),
    dtype="int32",
    name="encoder_inputs"
)

decoder_inputs = tf.keras.Input(
    shape=(DECODER_LEN,),
    dtype="int32",
    name="decoder_inputs"
)

embedding = tf.keras.layers.Embedding(
    input_dim=VOCAB_SIZE,
    output_dim=EMBED_DIM,
    mask_zero=True,
    name="embedding_50k"
)

encoder_emb = embedding(encoder_inputs)

decoder_emb = embedding(decoder_inputs)

encoder_lstm = tf.keras.layers.LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    name="encoder_lstm_50k"
)

encoder_outputs, encoder_h, encoder_c = encoder_lstm(
    encoder_emb
)

decoder_lstm = tf.keras.layers.LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    name="decoder_lstm_50k"
)

decoder_outputs, _, _ = decoder_lstm(
    decoder_emb,
    initial_state=[encoder_h, encoder_c]
)

attention = tf.keras.layers.AdditiveAttention(
    name="attention_50k"
)

context = attention(
    [decoder_outputs, encoder_outputs]
)

combined = tf.keras.layers.Concatenate(
    axis=-1,
    name="decoder_combined_50k"
)(
    [decoder_outputs, context]
)

output_layer = tf.keras.layers.Dense(
    VOCAB_SIZE,
    name="output_50k"
)

logits = output_layer(combined)

model = tf.keras.Model(
    inputs={
        "encoder_inputs": encoder_inputs,
        "decoder_inputs": decoder_inputs
    },
    outputs=logits,
    name="summarization_50k_fixed"
)

print()
model.summary()

# ------------------------------------------------------------
# 12. MASKED LOSS
# ------------------------------------------------------------

loss_function = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=True,
    reduction="none"
)

def masked_loss(y_true, y_pred):

    loss = loss_function(
        y_true,
        y_pred
    )

    mask = tf.cast(
        tf.not_equal(y_true, PAD_ID),
        tf.float32
    )

    loss = loss * mask

    return tf.reduce_sum(loss) / (
        tf.reduce_sum(mask) + 1e-7
    )

# ------------------------------------------------------------
# 13. COMPILE
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0005
    ),
    loss=masked_loss
)

print("Model compiled.")

# ------------------------------------------------------------
# 14. SAVE BEST MODEL TO DRIVE
# ------------------------------------------------------------

checkpoint = tf.keras.callbacks.ModelCheckpoint(
    MODEL_PATH,
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=1,
    restore_best_weights=True
)

# ------------------------------------------------------------
# 15. TRAIN
# ------------------------------------------------------------

print()
print("Starting 50K training...")
print()

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=[
        checkpoint,
        early_stop
    ]
)

# ------------------------------------------------------------
# 16. SAVE FINAL MODEL
# ------------------------------------------------------------

model.save(MODEL_PATH)

# ------------------------------------------------------------
# 17. SAVE TOKENIZER + ID CONFIG
# ------------------------------------------------------------

tokenizer_package = {
    "tokenizer": tokenizer,
    "vocab_size": VOCAB_SIZE,
    "pad_id": PAD_ID,
    "oov_id": OOV_ID,
    "start_id": START_ID,
    "end_id": END_ID,
    "encoder_len": ENCODER_LEN,
    "decoder_len": DECODER_LEN
}

with open(
    TOKENIZER_PATH,
    "wb"
) as f:
    pickle.dump(
        tokenizer_package,
        f,
        protocol=pickle.HIGHEST_PROTOCOL
    )

# ------------------------------------------------------------
# 18. VERIFY
# ------------------------------------------------------------

print()
print("==========================================")
print("REBUILD COMPLETE")
print("==========================================")

print("Model saved:")
print(MODEL_PATH)

print("Tokenizer saved:")
print(TOKENIZER_PATH)

print()
print("Model output:", model.output_shape)

print("PAD:", PAD_ID)
print("OOV:", OOV_ID)
print("START:", START_ID)
print("END:", END_ID)

print()
print("Checking tokenizer IDs:")

for word in [
    "the",
    "government",
    "project",
    "storm",
    "flooding"
]:
    raw_id = tokenizer.word_index.get(word)

    if raw_id is None:
        model_id = OOV_ID
    elif raw_id == 1:
        model_id = OOV_ID
    else:
        model_id = raw_id + 2

    print(
        word,
        "raw =", raw_id,
        "model =", model_id
    )

print()
print("Files exist:")
print("Model:", os.path.exists(MODEL_PATH))
print("Tokenizer:", os.path.exists(TOKENIZER_PATH))
```

```python
# ============================================================
# FIX 50K MODEL: DISABLE cuDNN MASK RESTRICTION
# Uses the tokenizer and encoded data already created above
# ============================================================

import os
import tensorflow as tf

VOCAB_SIZE = 50000
PAD_ID = 0

ENCODER_LEN = 500
DECODER_LEN = 66

EMBED_DIM = 128
LSTM_UNITS = 256
BATCH_SIZE = 32
EPOCHS = 3

DRIVE_DIR = "/content/drive/MyDrive/summarization_project"

MODEL_PATH = os.path.join(
    DRIVE_DIR,
    "summarization_50k_fixed.keras"
)

TOKENIZER_PATH = os.path.join(
    DRIVE_DIR,
    "tokenizer_50k_fixed.pkl"
)

# ------------------------------------------------------------
# INPUTS
# ------------------------------------------------------------

encoder_inputs = tf.keras.Input(
    shape=(ENCODER_LEN,),
    dtype="int32",
    name="encoder_inputs"
)

decoder_inputs = tf.keras.Input(
    shape=(DECODER_LEN,),
    dtype="int32",
    name="decoder_inputs"
)

# ------------------------------------------------------------
# SHARED 50K EMBEDDING
# ------------------------------------------------------------

embedding = tf.keras.layers.Embedding(
    input_dim=VOCAB_SIZE,
    output_dim=EMBED_DIM,
    mask_zero=True,
    name="embedding_50k"
)

encoder_emb = embedding(encoder_inputs)
decoder_emb = embedding(decoder_inputs)

# ------------------------------------------------------------
# ENCODER
#
# IMPORTANT:
# use_cudnn=False avoids the cuDNN mask assertion.
# ------------------------------------------------------------

encoder_lstm = tf.keras.layers.LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    use_cudnn=False,
    name="encoder_lstm_50k"
)

encoder_outputs, encoder_h, encoder_c = encoder_lstm(
    encoder_emb
)

# ------------------------------------------------------------
# DECODER
# ------------------------------------------------------------

decoder_lstm = tf.keras.layers.LSTM(
    LSTM_UNITS,
    return_sequences=True,
    return_state=True,
    use_cudnn=False,
    name="decoder_lstm_50k"
)

decoder_outputs, _, _ = decoder_lstm(
    decoder_emb,
    initial_state=[encoder_h, encoder_c]
)

# ------------------------------------------------------------
# ATTENTION
# ------------------------------------------------------------

attention = tf.keras.layers.AdditiveAttention(
    name="attention_50k"
)

context = attention(
    [decoder_outputs, encoder_outputs]
)

# ------------------------------------------------------------
# CONCATENATE
# ------------------------------------------------------------

combined = tf.keras.layers.Concatenate(
    axis=-1,
    name="decoder_combined_50k"
)(
    [decoder_outputs, context]
)

# ------------------------------------------------------------
# 50K OUTPUT
# ------------------------------------------------------------

output_layer = tf.keras.layers.Dense(
    VOCAB_SIZE,
    name="output_50k"
)

logits = output_layer(combined)

# ------------------------------------------------------------
# MODEL
# ------------------------------------------------------------

model = tf.keras.Model(
    inputs={
        "encoder_inputs": encoder_inputs,
        "decoder_inputs": decoder_inputs
    },
    outputs=logits,
    name="summarization_50k_fixed"
)

# ------------------------------------------------------------
# MASKED LOSS
# ------------------------------------------------------------

loss_function = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=True,
    reduction="none"
)

def masked_loss(y_true, y_pred):

    loss = loss_function(
        y_true,
        y_pred
    )

    mask = tf.cast(
        tf.not_equal(y_true, PAD_ID),
        tf.float32
    )

    loss = loss * mask

    return tf.reduce_sum(loss) / (
        tf.reduce_sum(mask) + 1e-7
    )

# ------------------------------------------------------------
# COMPILE
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0005
    ),
    loss=masked_loss
)

print("Model created successfully")
print("Output:", model.output_shape)

# ------------------------------------------------------------
# DATASET
# ------------------------------------------------------------

train_dataset = tf.data.Dataset.from_tensor_slices(
    (
        {
            "encoder_inputs": X_train,
            "decoder_inputs": decoder_inputs_train
        },
        decoder_targets_train
    )
)

train_dataset = (
    train_dataset
    .shuffle(20000)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

val_dataset = tf.data.Dataset.from_tensor_slices(
    (
        {
            "encoder_inputs": X_val,
            "decoder_inputs": decoder_inputs_val
        },
        decoder_targets_val
    )
)

val_dataset = (
    val_dataset
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

# ------------------------------------------------------------
# CHECKPOINT
# ------------------------------------------------------------

checkpoint = tf.keras.callbacks.ModelCheckpoint(
    MODEL_PATH,
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=1,
    restore_best_weights=True
)

# ------------------------------------------------------------
# TRAIN
# ------------------------------------------------------------

print()
print("Starting training...")
print()

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=[
        checkpoint,
        early_stop
    ]
)

# ------------------------------------------------------------
# SAVE FINAL MODEL
# ------------------------------------------------------------

model.save(MODEL_PATH)

print()
print("==========================================")
print("TRAINING FINISHED")
print("==========================================")
print("Model saved:")
print(MODEL_PATH)

print("Exists:", os.path.exists(MODEL_PATH))

if os.path.exists(MODEL_PATH):
    print(
        "Size:",
        round(os.path.getsize(MODEL_PATH) / (1024**2), 2),
        "MB"
    )
```

```python
import os
import pickle
import numpy as np
import tensorflow as tf

# ============================================================
# LOAD
# ============================================================

MODEL_PATH = "/content/drive/MyDrive/summarization_project/summarization_50k_fixed.keras"
TOKENIZER_PATH = "/content/drive/MyDrive/summarization_project/tokenizer_corrected.pkl"

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

print("Model loaded")
print("Tokenizer loaded")

# ============================================================
# SPECIAL IDS
# ============================================================

PAD_ID = 0
OOV_ID = 1
START_ID = 2
END_ID = 3

MAX_INPUT_LEN = 500
MAX_OUTPUT_LEN = 66

print()
print("Special IDs:")
print("PAD  =", PAD_ID)
print("OOV  =", OOV_ID)
print("START=", START_ID)
print("END  =", END_ID)

# ============================================================
# GET TRAINED LAYERS
# ============================================================

embedding_layer = model.get_layer("embedding")
encoder_lstm = model.get_layer("encoder_lstm_50k")
decoder_lstm = model.get_layer("decoder_lstm_50k")
attention_layer = model.get_layer("attention_50k")
output_layer = model.get_layer("output_layer_50k")

print()
print("Layers found:")
print("Embedding:", embedding_layer.name)
print("Encoder:", encoder_lstm.name)
print("Decoder:", decoder_lstm.name)
print("Attention:", attention_layer.name)
print("Output:", output_layer.name)

# ============================================================
# ENCODER
# ============================================================

def encode_document(text):

    seq = tokenizer.texts_to_sequences([text])[0]

    # Keep only IDs usable by the model/tokenizer setup.
    seq = [int(x) for x in seq if 0 <= int(x) < 50000]

    # Right padding.
    seq = seq[:MAX_INPUT_LEN]

    padded = np.zeros((1, MAX_INPUT_LEN), dtype=np.int32)
    padded[0, :len(seq)] = seq

    return padded


# ============================================================
# AUTOREGRESSIVE DECODER
# ============================================================

def generate_summary(text, max_len=40):

    encoder_input = encode_document(text)

    # --------------------------------------------------------
    # Encoder
    # --------------------------------------------------------

    encoder_emb = embedding_layer(encoder_input)

    encoder_output, encoder_h, encoder_c = encoder_lstm(
        encoder_emb,
        training=False
    )

    # --------------------------------------------------------
    # Start decoder
    # --------------------------------------------------------

    current_id = START_ID

    generated_ids = []

    for step in range(max_len):

        decoder_input = tf.constant(
            [[current_id]],
            dtype=tf.int32
        )

        decoder_emb = embedding_layer(decoder_input)

        decoder_output, decoder_h, decoder_c = decoder_lstm(
            decoder_emb,
            initial_state=[encoder_h, encoder_c],
            training=False
        )

        # ----------------------------------------------------
        # Attention
        # ----------------------------------------------------

        context = attention_layer(
            [decoder_output, encoder_output],
            training=False
        )

        # ----------------------------------------------------
        # Combine decoder + attention
        # ----------------------------------------------------

        combined = tf.concat(
            [decoder_output, context],
            axis=-1
        )

        logits = output_layer(
            combined,
            training=False
        )

        # Shape:
        # (1, 1, 50000)

        logits = logits[:, -1, :]

        # ----------------------------------------------------
        # Prevent invalid / unused IDs
        # ----------------------------------------------------

        logits = tf.tensor_scatter_nd_update(
            logits,
            indices=tf.constant(
                [[0, 0]]
            ),
            updates=tf.constant(
                [-1e9],
                dtype=logits.dtype
            )
        )

        # IDs >= tokenizer.num_words are not represented by
        # the current tokenizer configuration.
        if tokenizer.num_words is not None:

            limit = min(
                int(tokenizer.num_words),
                int(logits.shape[-1])
            )

            if limit < int(logits.shape[-1]):

                mask = tf.concat(
                    [
                        tf.ones((1, limit), dtype=tf.bool),
                        tf.zeros(
                            (1, int(logits.shape[-1]) - limit),
                            dtype=tf.bool
                        )
                    ],
                    axis=1
                )

                logits = tf.where(
                    mask,
                    logits,
                    tf.ones_like(logits) * -1e9
                )

        # ----------------------------------------------------
        # Greedy prediction
        # ----------------------------------------------------

        next_id = int(
            tf.argmax(
                logits,
                axis=-1
            )[0].numpy()
        )

        generated_ids.append(next_id)

        # ----------------------------------------------------
        # Stop at END
        # ----------------------------------------------------

        if next_id == END_ID:
            break

        current_id = next_id

        # Next decoder step should use the NEW decoder state.
        encoder_h = decoder_h
        encoder_c = decoder_c

    # --------------------------------------------------------
    # Convert IDs -> words
    # --------------------------------------------------------

    reverse_word_index = {
        int(v): k
        for k, v in tokenizer.word_index.items()
    }

    words = []

    for idx in generated_ids:

        idx = int(idx)

        if idx == START_ID:
            continue

        if idx == END_ID:
            break

        if idx == PAD_ID:
            continue

        word = reverse_word_index.get(
            idx,
            "<OOV>"
        )

        words.append(word)

    return " ".join(words), generated_ids


# ============================================================
# TEST
# ============================================================

test_document = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

summary, ids = generate_summary(
    test_document,
    max_len=40
)

print()
print("=" * 70)
print("AUTOREGRESSIVE TEST")
print("=" * 70)

print()
print("DOCUMENT:")
print(test_document.strip())

print()
print("GENERATED SUMMARY:")
print(summary)

print()
print("GENERATED IDS:")
print(ids)

print()
print("GENERATED LENGTH:")
print(len(ids))
```

```python
import tensorflow as tf
import numpy as np
import pickle

MODEL_PATH = "/content/drive/MyDrive/summarization_project/summarization_50k_fixed.keras"
TOKENIZER_PATH = "/content/drive/MyDrive/summarization_project/tokenizer_corrected.pkl"

print("=" * 70)
print("LOADING 50K MODEL FOR PROPER INFERENCE")
print("=" * 70)

model = tf.keras.models.load_model(MODEL_PATH, compile=False)

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

print("Model loaded")
print("Tokenizer loaded")

# ------------------------------------------------------------
# IMPORTANT CONTROL IDS
# ------------------------------------------------------------

PAD_ID = 0
OOV_ID = 1

# These were used during training as decoder control tokens.
START_ID = 2
END_ID = 3

print("\nCONTROL IDS")
print("PAD   =", PAD_ID)
print("OOV   =", OOV_ID)
print("START =", START_ID)
print("END   =", END_ID)

print("\nTokenizer's normal interpretation:")
print("2 ->", tokenizer.index_word.get(2))
print("3 ->", tokenizer.index_word.get(3))

# ------------------------------------------------------------
# TOKENIZE INPUT
# ------------------------------------------------------------

def encode_document(text):
    ids = tokenizer.texts_to_sequences([text])[0]

    # Keep IDs within encoder vocabulary range.
    ids = [x if x < 50000 else OOV_ID for x in ids]

    # Right padding.
    ids = ids[:500]
    ids = ids + [PAD_ID] * (500 - len(ids))

    return np.array([ids], dtype=np.int32)


# ------------------------------------------------------------
# MANUAL DECODER ID -> WORD
# ------------------------------------------------------------

def decode_model_ids(ids):
    words = []

    for idx in ids:

        idx = int(idx)

        # Padding
        if idx == PAD_ID:
            continue

        # START is a control token
        if idx == START_ID:
            continue

        # END is a control token
        if idx == END_ID:
            break

        # OOV
        if idx == OOV_ID:
            words.append("<OOV>")
            continue

        # Normal tokenizer vocabulary
        word = tokenizer.index_word.get(idx)

        if word is None:
            words.append("<UNK>")
        else:
            words.append(word)

    return " ".join(words)


# ------------------------------------------------------------
# AUTOREGRESSIVE DECODING
# ------------------------------------------------------------

def generate_summary(text, max_length=66):

    encoder_input = encode_document(text)

    # Decoder starts with START ID = 2.
    decoder_ids = [START_ID]

    for step in range(max_length - 1):

        # Current decoder sequence
        current = decoder_ids[:max_length]

        # Right-pad decoder input to 66.
        padded_decoder = current + [PAD_ID] * (max_length - len(current))

        padded_decoder = np.array(
            [padded_decoder],
            dtype=np.int32
        )

        # Predict entire decoder sequence.
        prediction = model.predict(
            {
                "encoder_inputs": encoder_input,
                "decoder_inputs": padded_decoder
            },
            verbose=0
        )

        # Take prediction corresponding to the newest decoder position.
        position = len(current) - 1

        next_id = int(
            np.argmax(
                prediction[0, position]
            )
        )

        decoder_ids.append(next_id)

        # Stop at END.
        if next_id == END_ID:
            break

    return decoder_ids, decode_model_ids(decoder_ids)


# ------------------------------------------------------------
# TEST DOCUMENT
# ------------------------------------------------------------

document = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

ids, summary = generate_summary(document)

print("\n" + "=" * 70)
print("INPUT")
print("=" * 70)
print(document.strip())

print("\n" + "=" * 70)
print("GENERATED MODEL IDS")
print("=" * 70)
print(ids)

print("\n" + "=" * 70)
print("GENERATED SUMMARY")
print("=" * 70)
print(summary)

print("\n" + "=" * 70)
print("ID / WORD CHECK")
print("=" * 70)

for idx in ids:
    print(
        f"{idx:6d} -> "
        f"{tokenizer.index_word.get(idx, '<CONTROL/UNKNOWN>')}"
    )
```

```python
# ============================================================
# TEST CLEAN 50K MODEL
# ============================================================

import os
import pickle
import numpy as np
import tensorflow as tf

DRIVE_DIR = "/content/drive/MyDrive/summarization_project"

MODEL_PATH = os.path.join(
    DRIVE_DIR,
    "summarization_50k_clean.keras"
)

TOKENIZER_PATH = os.path.join(
    DRIVE_DIR,
    "tokenizer_clean_50k.pkl"
)

MAX_INPUT_LEN = 500
MAX_TARGET_LEN = 66

START_ID = 2
END_ID = 3
PAD_ID = 0
OOV_ID = 1

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

with open(TOKENIZER_PATH, "rb") as f:
    tokenizer = pickle.load(f)

print("Model loaded")
print("Tokenizer loaded")

print()
print("MODEL OUTPUT:", model.output.shape)
print("TOKENIZER VOCABULARY:", len(tokenizer.word_index))
print("TOKENIZER NUM WORDS:", tokenizer.num_words)

# ------------------------------------------------------------
# VERIFY SPECIAL IDS
# ------------------------------------------------------------

print()
print("SPECIAL IDS")
print("PAD   =", PAD_ID)
print("OOV   =", OOV_ID)
print("START =", tokenizer.word_index.get("<START>"))
print("END   =", tokenizer.word_index.get("<END>"))

assert tokenizer.word_index["<START>"] == START_ID
assert tokenizer.word_index["<END>"] == END_ID

# ------------------------------------------------------------
# WORD -> ID
# ------------------------------------------------------------

def encode_document(text):
    ids = tokenizer.texts_to_sequences([text])[0]

    ids = ids[:MAX_INPUT_LEN]

    if len(ids) < MAX_INPUT_LEN:
        ids = ids + [PAD_ID] * (
            MAX_INPUT_LEN - len(ids)
        )

    return np.array([ids], dtype=np.int32)


# ------------------------------------------------------------
# ID -> WORD
# ------------------------------------------------------------

index_word = tokenizer.index_word

def decode_ids(ids):
    words = []

    for token_id in ids:

        token_id = int(token_id)

        if token_id == START_ID:
            continue

        if token_id == END_ID:
            break

        if token_id == PAD_ID:
            continue

        word = index_word.get(token_id)

        if word is None:
            continue

        if word in ["<START>", "<END>"]:
            continue

        words.append(word)

    return " ".join(words)


# ------------------------------------------------------------
# AUTOREGRESSIVE DECODING
# ------------------------------------------------------------

def generate_summary(text):

    encoder_input = encode_document(text)

    decoder_input = np.zeros(
        (1, MAX_TARGET_LEN),
        dtype=np.int32
    )

    decoder_input[0, 0] = START_ID

    generated_ids = []

    for step in range(MAX_TARGET_LEN - 1):

        prediction = model.predict(
            {
                "encoder_inputs": encoder_input,
                "decoder_inputs": decoder_input
            },
            verbose=0
        )

        next_id = int(
            np.argmax(
                prediction[0, step]
            )
        )

        if next_id == END_ID:
            break

        if next_id == PAD_ID:
            break

        generated_ids.append(next_id)

        decoder_input[0, step + 1] = next_id

    return generated_ids


# ============================================================
# TEST DOCUMENT
# ============================================================

document = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

print()
print("=" * 70)
print("TEST DOCUMENT")
print("=" * 70)

print(document.strip())

# ------------------------------------------------------------
# TOKENIZATION TEST
# ------------------------------------------------------------

input_ids = encode_document(document)[0]

print()
print("FIRST INPUT IDS:")
print(input_ids[:30])

print()
print("FIRST INPUT WORDS:")

for token_id in input_ids[:30]:

    if token_id == PAD_ID:
        continue

    print(
        token_id,
        "->",
        index_word.get(int(token_id), "<UNKNOWN>")
    )

# ------------------------------------------------------------
# GENERATE
# ------------------------------------------------------------

generated_ids = generate_summary(document)

generated_text = decode_ids(generated_ids)

print()
print("=" * 70)
print("GENERATED IDS")
print("=" * 70)

print(generated_ids)

print()
print("=" * 70)
print("GENERATED SUMMARY")
print("=" * 70)

print(generated_text)

print()
print("Generated length:", len(generated_ids))
```

```python
# ============================================================
# KARTHIK TRAINED MODEL
# PRETRAINED BART-LARGE-XSUM + XSUM FINE-TUNING
#
# IMPORTANT:
# - DO NOT use tokenizer_clean_50k.pkl
# - DO NOT use the old LSTM model
# - BART's own tokenizer is used
# - Encoder + Decoder are fine-tuned together
# - Designed for Tesla T4 14.56 GB
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

!pip -q install -U transformers datasets evaluate accelerate sentencepiece rouge_score

# ============================================================
# 2. IMPORTS
# ============================================================

import os
import gc
import torch

from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
    DataCollatorForSeq2Seq,
    Seq2SeqTrainingArguments,
    Seq2SeqTrainer,
    EarlyStoppingCallback
)

import evaluate

# ============================================================
# 3. CLEAR GPU MEMORY
# ============================================================

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

print("=" * 70)
print("GPU CHECK")
print("=" * 70)

print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

    total_memory = torch.cuda.get_device_properties(0).total_memory
    print(
        "GPU memory:",
        round(total_memory / (1024 ** 3), 2),
        "GB"
    )

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using:", device)

# ============================================================
# 4. PATHS
# ============================================================

DRIVE_DIR = "/content/drive/MyDrive/summarization_project"

FINAL_MODEL_DIR = os.path.join(
    DRIVE_DIR,
    "karthik_trained_model"
)

BEST_MODEL_DIR = os.path.join(
    DRIVE_DIR,
    "karthik_trained_model_best"
)

os.makedirs(DRIVE_DIR, exist_ok=True)

# ============================================================
# 5. PRETRAINED MODEL
# ============================================================

MODEL_NAME = "facebook/bart-large-xsum"

print()
print("=" * 70)
print("LOADING PRETRAINED BART-LARGE-XSUM")
print("=" * 70)

print("Model:", MODEL_NAME)

# ============================================================
# 6. BART TOKENIZER
#
# IMPORTANT:
# This is BART's tokenizer.
#
# We DO NOT load:
# tokenizer_clean_50k.pkl
#
# We DO NOT create:
# custom 50K tokenizer
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    use_fast=True
)

print()
print("BART TOKENIZER LOADED")
print("Tokenizer class:", tokenizer.__class__.__name__)
print("Vocabulary size:", len(tokenizer))

print()
print("BART SPECIAL TOKENS")
print("PAD:", tokenizer.pad_token, tokenizer.pad_token_id)
print("BOS:", tokenizer.bos_token, tokenizer.bos_token_id)
print("EOS:", tokenizer.eos_token, tokenizer.eos_token_id)

# ============================================================
# 7. LOAD PRETRAINED BART
# ============================================================

print()
print("=" * 70)
print("LOADING BART MODEL")
print("=" * 70)

model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16
)

print("Model loaded.")

# ============================================================
# 8. MEMORY-SAVING SETTINGS
# ============================================================

# Gradient checkpointing saves GPU memory during training.
model.gradient_checkpointing_enable()

# Required when using gradient checkpointing with some
# Transformer models.
model.config.use_cache = False

print("Gradient checkpointing: ENABLED")
print("use_cache:", model.config.use_cache)

# ============================================================
# 9. LOAD XSUM DATASET
# ============================================================

print()
print("=" * 70)
print("LOADING XSUM DATASET")
print("=" * 70)

dataset = load_dataset("EdinburghNLP/xsum")

print(dataset)

print()
print("Full train:", len(dataset["train"]))
print("Full validation:", len(dataset["validation"]))

# ============================================================
# 10. USE YOUR PROJECT SIZE
#
# 20,000 train
# 2,000 validation
#
# Same scale you used with the LSTM experiment.
# ============================================================

TRAIN_SIZE = 20000
VAL_SIZE = 2000

train_dataset = dataset["train"].select(
    range(TRAIN_SIZE)
)

val_dataset = dataset["validation"].select(
    range(VAL_SIZE)
)

print()
print("Training examples:", len(train_dataset))
print("Validation examples:", len(val_dataset))

# ============================================================
# 11. TOKENIZATION
# ============================================================

MAX_SOURCE_LENGTH = 512
MAX_TARGET_LENGTH = 64

print()
print("=" * 70)
print("TOKENIZING XSUM WITH BART TOKENIZER")
print("=" * 70)

def preprocess_function(examples):

    # XSum document
    inputs = examples["document"]

    # XSum one-sentence summary
    targets = examples["summary"]

    model_inputs = tokenizer(
        inputs,
        max_length=MAX_SOURCE_LENGTH,
        truncation=True,
        padding=False
    )

    labels = tokenizer(
        text_target=targets,
        max_length=MAX_TARGET_LENGTH,
        truncation=True,
        padding=False
    )

    model_inputs["labels"] = labels["input_ids"]

    return model_inputs


tokenized_train = train_dataset.map(
    preprocess_function,
    batched=True,
    remove_columns=train_dataset.column_names,
    desc="Tokenizing training data"
)

tokenized_val = val_dataset.map(
    preprocess_function,
    batched=True,
    remove_columns=val_dataset.column_names,
    desc="Tokenizing validation data"
)

print()
print("Tokenization complete.")

print("Train examples:", len(tokenized_train))
print("Validation examples:", len(tokenized_val))

# ============================================================
# 12. DATA COLLATOR
# ============================================================

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
    padding=True,
    pad_to_multiple_of=8
)

print()
print("Data collator created.")

# ============================================================
# 13. ROUGE
# ============================================================

rouge = evaluate.load("rouge")

def compute_metrics(eval_preds):

    predictions, labels = eval_preds

    # Some versions return a tuple.
    if isinstance(predictions, tuple):
        predictions = predictions[0]

    # Decode predictions
    decoded_predictions = tokenizer.batch_decode(
        predictions,
        skip_special_tokens=True
    )

    # Replace -100 with PAD before decoding labels
    labels = torch.tensor(labels)

    labels = torch.where(
        labels == -100,
        torch.tensor(tokenizer.pad_token_id),
        labels
    )

    decoded_labels = tokenizer.batch_decode(
        labels,
        skip_special_tokens=True
    )

    # Clean whitespace
    decoded_predictions = [
        pred.strip()
        for pred in decoded_predictions
    ]

    decoded_labels = [
        label.strip()
        for label in decoded_labels
    ]

    scores = rouge.compute(
        predictions=decoded_predictions,
        references=decoded_labels,
        use_stemmer=True
    )

    return {
        "rouge1": round(scores["rouge1"] * 100, 4),
        "rouge2": round(scores["rouge2"] * 100, 4),
        "rougeL": round(scores["rougeL"] * 100, 4)
    }

# ============================================================
# 14. TRAINING ARGUMENTS
#
# T4 MEMORY SETTINGS
# ============================================================

training_args = Seq2SeqTrainingArguments(

    output_dir=BEST_MODEL_DIR,

    # --------------------------------------------------------
    # EPOCHS
    # --------------------------------------------------------
    num_train_epochs=3,

    # --------------------------------------------------------
    # SMALL GPU BATCH
    # --------------------------------------------------------
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,

    # Accumulate 8 batches before optimizer update.
    gradient_accumulation_steps=8,

    # --------------------------------------------------------
    # LEARNING RATE
    # --------------------------------------------------------
    learning_rate=3e-5,

    weight_decay=0.01,

    warmup_ratio=0.05,

    # --------------------------------------------------------
    # FP16
    # --------------------------------------------------------
    fp16=True,
    bf16=False,

    # --------------------------------------------------------
    # EVALUATION
    # --------------------------------------------------------
    eval_strategy="steps",
    eval_steps=1000,

    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------
    save_strategy="steps",
    save_steps=1000,
    save_total_limit=2,

    # --------------------------------------------------------
    # BEST MODEL
    # --------------------------------------------------------
    load_best_model_at_end=True,
    metric_for_best_model="rougeL",
    greater_is_better=True,

    # --------------------------------------------------------
    # GENERATION
    # --------------------------------------------------------
    predict_with_generate=True,

    generation_max_length=64,
    generation_num_beams=4,

    # --------------------------------------------------------
    # LOGGING
    # --------------------------------------------------------
    logging_steps=100,

    # --------------------------------------------------------
    # PERFORMANCE
    # --------------------------------------------------------
    dataloader_num_workers=2,

    # --------------------------------------------------------
    # REPORTING
    # --------------------------------------------------------
    report_to="none",

    # --------------------------------------------------------
    # LABEL HANDLING
    # --------------------------------------------------------
    label_smoothing_factor=0.0,

    # --------------------------------------------------------
    # MEMORY
    # --------------------------------------------------------
    gradient_checkpointing=True,

    # --------------------------------------------------------
    # SAVE SAFETENSORS
    # --------------------------------------------------------
    save_safetensors=True
)

# ============================================================
# 15. TRAINER
# ============================================================

trainer = Seq2SeqTrainer(

    model=model,

    args=training_args,

    train_dataset=tokenized_train,

    eval_dataset=tokenized_val,

    tokenizer=tokenizer,

    data_collator=data_collator,

    compute_metrics=compute_metrics,

    callbacks=[
        EarlyStoppingCallback(
            early_stopping_patience=2
        )
    ]
)

# ============================================================
# 16. SHOW TRAINING INFORMATION
# ============================================================

print()
print("=" * 70)
print("TRAINING CONFIGURATION")
print("=" * 70)

print("Model:", MODEL_NAME)
print("Train:", TRAIN_SIZE)
print("Validation:", VAL_SIZE)
print("Epochs:", 3)
print("Batch size:", 1)
print("Gradient accumulation:", 8)
print("Effective batch size:", 8)
print("Learning rate:", 3e-5)
print("FP16:", True)
print("Max source length:", MAX_SOURCE_LENGTH)
print("Max target length:", MAX_TARGET_LENGTH)
print("Beam size:", 4)

# ============================================================
# 17. TRAIN
# ============================================================

print()
print("=" * 70)
print("STARTING BART XSUM FINE-TUNING")
print("=" * 70)

train_result = trainer.train()

print()
print("=" * 70)
print("TRAINING FINISHED")
print("=" * 70)

# ============================================================
# 18. SAVE FINAL MODEL
# ============================================================

print()
print("=" * 70)
print("SAVING KARTHIK TRAINED MODEL")
print("=" * 70)

trainer.save_model(FINAL_MODEL_DIR)

tokenizer.save_pretrained(FINAL_MODEL_DIR)

print()
print("MODEL SAVED:")
print(FINAL_MODEL_DIR)

print()
print("Tokenizer saved with model:")
print(FINAL_MODEL_DIR)

# ============================================================
# 19. FINAL EVALUATION
# ============================================================

print()
print("=" * 70)
print("FINAL VALIDATION")
print("=" * 70)

metrics = trainer.evaluate(
    max_length=MAX_TARGET_LENGTH,
    num_beams=4
)

for key, value in metrics.items():

    if isinstance(value, float):
        print(
            f"{key}: {value:.4f}"
        )
    else:
        print(
            f"{key}: {value}"
        )

# ============================================================
# 20. TEST SUMMARY GENERATION
# ============================================================

test_document = """
The government announced a new program to improve public transportation
across the city. The program will add electric buses, upgrade existing
stations, and introduce more frequent services during peak hours.
Officials said the project is expected to reduce traffic congestion
and lower pollution over the next five years.
"""

print()
print("=" * 70)
print("TEST DOCUMENT")
print("=" * 70)

print(test_document)

# Put model in evaluation mode
model.eval()

inputs = tokenizer(
    test_document,
    return_tensors="pt",
    max_length=MAX_SOURCE_LENGTH,
    truncation=True
)

# Move input tensors to GPU
inputs = {
    key: value.to(model.device)
    for key, value in inputs.items()
}

print()
print("=" * 70)
print("GENERATING SUMMARY")
print("=" * 70)

with torch.no_grad():

    generated_ids = model.generate(
        **inputs,

        max_length=64,
        min_length=8,

        num_beams=4,

        early_stopping=True,

        no_repeat_ngram_size=3,

        length_penalty=1.0,

        repetition_penalty=1.05
    )

generated_summary = tokenizer.decode(
    generated_ids[0],
    skip_special_tokens=True
)

print()
print("SUMMARY:")
print(generated_summary)

# ============================================================
# 21. GPU MEMORY
# ============================================================

if torch.cuda.is_available():

    allocated = torch.cuda.memory_allocated() / (1024 ** 3)
    reserved = torch.cuda.memory_reserved() / (1024 ** 3)

    print()
    print("=" * 70)
    print("GPU MEMORY")
    print("=" * 70)

    print(
        "Allocated:",
        round(allocated, 3),
        "GB"
    )

    print(
        "Reserved:",
        round(reserved, 3),
        "GB"
    )

# ============================================================
# 22. FINAL CHECK
# ============================================================

print()
print("=" * 70)
print("FINAL RESULT")
print("=" * 70)

print("Model directory:")
print(FINAL_MODEL_DIR)

print(
    "Exists:",
    os.path.exists(FINAL_MODEL_DIR)
)

print()
print("IMPORTANT:")
print("This model uses BART's tokenizer.")
print("The old 50K tokenizer was NOT used.")
print("The old LSTM model was NOT used.")
print("Both BART encoder and decoder were fine-tuned.")
print()
print("DONE.")
```

```python
# ============================================================
# FINAL SUMMARIZATION FUNCTION
# ============================================================

def summarize(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        max_length=16384,
        truncation=True
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=100,
            num_beams=4,
            no_repeat_ngram_size=3,
            length_penalty=1.0
        )

    return tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )


# ============================================================
# TRY YOUR OWN TEXT HERE
# ============================================================

text = """
Artificial intelligence is changing many industries by helping
organizations automate repetitive tasks, analyze large amounts of
information and provide faster services. Businesses are using AI for
customer support, fraud detection, recommendation systems and data
analysis. At the same time, organizations need to consider privacy,
security, accuracy and the impact of automation on employees. Experts
recommend that companies use AI responsibly and maintain human
oversight for important decisions.
"""

summary = summarize(text)

print("=" * 60)
print("ORIGINAL")
print("=" * 60)
print(text.strip())

print()
print("=" * 60)
print("SUMMARY")
print("=" * 60)
print(summary)
```

```python
# ============================================================
# BART LONG PARAGRAPH TEST
# ============================================================

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = "facebook/bart-large-cnn"

print("=" * 60)
print("LOADING BART")
print("=" * 60)

bart_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
bart_model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

bart_model = bart_model.to(device)
bart_model.eval()

print("Model loaded")
print("Device:", device)

# ============================================================
# LONG TEST PARAGRAPH
# ============================================================

text = """
The government has announced a major five-year plan to modernize public
transportation across the city. The project will introduce hundreds of
electric buses, replace older diesel vehicles, upgrade existing bus
stations and improve accessibility for passengers with disabilities.
Officials also plan to increase the frequency of bus and train services
during morning and evening peak hours so that passengers spend less time
waiting at stations.

According to the transportation department, the project is expected to
reduce the number of private vehicles on major roads and help ease
traffic congestion. Officials also expect the use of electric buses to
reduce air pollution and greenhouse gas emissions across the city.
Construction and installation work will begin next year, with the first
phase expected to be completed within eighteen months.

The government said the project will be funded through a combination of
municipal funds, state support and additional transportation grants.
Officials added that progress will be reviewed every six months and that
future phases will be adjusted according to passenger demand and the
results of the first phase.
"""

# ============================================================
# TOKENIZE
# ============================================================

inputs = bart_tokenizer(
    text,
    return_tensors="pt",
    max_length=1024,
    truncation=True
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}

print()
print("Input tokens:", inputs["input_ids"].shape[1])

# ============================================================
# GENERATE SUMMARY
# ============================================================

with torch.no_grad():
    output_ids = bart_model.generate(
        **inputs,
        max_new_tokens=100,
        num_beams=4,
        no_repeat_ngram_size=3,
        length_penalty=1.0,
        early_stopping=True
    )

summary = bart_tokenizer.decode(
    output_ids[0],
    skip_special_tokens=True
)

# ============================================================
# DISPLAY RESULT
# ============================================================

print()
print("=" * 60)
print("ORIGINAL")
print("=" * 60)
print(text.strip())

print()
print("=" * 60)
print("BART LONG-PARAGRAPH SUMMARY")
print("=" * 60)
print(summary)

print()
print("Summary tokens:", output_ids.shape[1])
```
